# Step 9: Assess Missing Values

This step identifies the number of missing values in each column. Detecting missing values is an important part of data quality assessment and helps determine whether preprocessing is required.

```python
df.isnull().sum()
```

# Step 10: Calculate Missing Value Percentage

This step calculates the percentage of missing values in each column. Expressing missing values as percentages provides a clearer understanding of data completeness and helps prioritize columns that may require attention.

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage.sort_values(ascending=False)
```

# Step 11: Handle Missing Values

A missing value assessment revealed that the dataset contains only one missing value in the **Traveller Type** column. Since the missing value represents less than 0.1% of the dataset, the corresponding record is removed to maintain data quality without significantly affecting the analysis.

```python
df = df.dropna(subset=["traveller_type"])
df.isnull().sum()
```

# Step 12: Export the Cleaned Dataset

After completing the data cleaning process, the cleaned dataset is exported as a new CSV file. This file will be used as the primary data source for building interactive dashboards in Tableau.

The exported dataset preserves the original data structure while incorporating the necessary cleaning performed during preprocessing.

```python
df.to_csv("ba_reviews_cleaned.csv", index=False)
```
