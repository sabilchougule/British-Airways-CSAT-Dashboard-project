# Step 12: Inspect Unique Values

This step examines the unique values present in each column of the Countries dataset. It helps verify data consistency and identify any unexpected or inconsistent entries before using the dataset in Tableau.

```python
for col in countries.columns:
    print(f"\nUnique values in {col}:")
    print(countries[col].unique())
```

## Observation

The unique value analysis confirms that the Countries dataset is consistent and well-structured.

Key observations:
- The dataset contains **251 unique countries**.
- There are **250 unique country codes**, with one missing code corresponding to **Sark**.
- The dataset includes **7 continents** and **23 geographical regions**.
- No duplicate or inconsistent category values were identified.

The dataset is suitable for establishing geographical relationships in Tableau.
